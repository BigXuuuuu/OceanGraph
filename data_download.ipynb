```python
import torch
import netCDF4 as nc
from netCDF4 import Dataset
import numpy as np

# 指定数据路径
file_path = "/root/autodl-tmp/cmems_mod_glo_phy_my_0.083deg_P1D-m_multi-vars_180.00W-179.92E_80.00S-90.00N_0.49-2.65m_2021-01-06-2021-06-30.nc"

# 读取 NetCDF 文件
with nc.Dataset(file_path, 'r') as dataset:
    # 获取维度数据并转为 NumPy 数组
    lat = dataset.variables['latitude'][:]
    lon = dataset.variables['longitude'][:]
    time = dataset.variables['time'][:]
    level = dataset.variables['depth'][:]
    
    # 列出你要处理的变量
    variables_to_process = ['siconc', 'sithick', 'so', 'thetao', 'zos', 'usi', 'vsi']
    
    # 用来存储移动到 GPU 的张量
    gpu_data = {}
    
    # 遍历并处理每个变量
    for var in variables_to_process:
        if var in dataset.variables:
            # 读取变量数据并转为 NumPy 数组
            var_data = dataset.variables[var][:]  # (time, lat, lon)
            # 将 NumPy 数组转换为 PyTorch 张量，并移动到 GPU
            var_tensor = torch.tensor(var_data, dtype=torch.float32).cuda()
            # 存储到 gpu_data 字典中
            gpu_data[var] = var_tensor

# 示例：打印每个变量的张量形状
for var, tensor in gpu_data.items():
    print(f"{var} tensor shape: {tensor.shape}")

# 3. 执行降采样操作
def coarsen_data(data, factor):
    """在 GPU 上执行降采样操作"""
    # 使用 unfold 和 mean 在 lat 和 lon 维度进行降采样
    data_coarsened = data.unfold(2, factor, factor).unfold(3, factor, factor).mean(dim=(-1, -2))
    return data_coarsened

coarsen_factor = 4
for var in gpu_data:
    gpu_data[var] = coarsen_data(gpu_data[var], coarsen_factor)

# 4. 处理批次的时间维度
batch_size = 14
num_days = time.shape[0]
num_batches = num_days // batch_size
truncated_num_days = num_batches * batch_size

# 截断时间维度以适应完整批次
time_tensor = torch.tensor(time[:truncated_num_days], dtype=torch.float32).cuda()
for var in gpu_data:
    gpu_data[var] = gpu_data[var][:truncated_num_days]

# 重塑数据以创建批次
for var in gpu_data:
    gpu_data[var] = gpu_data[var].reshape((num_batches, batch_size, *gpu_data[var].shape[1:]))

# 5. 生成新的均匀间隔的经纬度坐标
lat_tensor = torch.tensor(lat, dtype=torch.float32).cuda()
lon_tensor = torch.tensor(lon, dtype=torch.float32).cuda()
new_lat = torch.linspace(lat_tensor.min(), lat_tensor.max(), lat_tensor.size(0)).cuda()
new_lon = torch.linspace(lon_tensor.min(), lon_tensor.max(), lon_tensor.size(0)).cuda()

# 对数据集进行插值（重采样）
def interp_data(data, new_lat, new_lon):
    """假设有插值函数, 这里用 nearest-neighbor 方式"""
    # 示例：这里只是返回原始数据，实际插值逻辑应根据需求实现
    return data

for var in gpu_data:
    gpu_data[var] = interp_data(gpu_data[var], new_lat, new_lon)

# 6. 将处理后的张量数据保存为 netCDF 文件
def save_to_netcdf(filename, data_dict, lat, lon, time):
    with Dataset(filename, 'w', format='NETCDF4') as ncfile:
        # 创建维度
        ncfile.createDimension('lat', lat.size(0))
        ncfile.createDimension('lon', lon.size(0))
        ncfile.createDimension('time', time.size(0))

        # 创建变量
        latitudes = ncfile.createVariable('latitude', 'f4', ('lat',))
        longitudes = ncfile.createVariable('longitude', 'f4', ('lon',))
        times = ncfile.createVariable('time', 'f4', ('time',))

        # 写入数据
        latitudes[:] = lat.cpu().numpy()
        longitudes[:] = lon.cpu().numpy()
        times[:] = time.cpu().numpy()

        # 创建数据变量
        for var, data in data_dict.items():
            data_var = ncfile.createVariable(var, 'f4', ('time', 'lat', 'lon'))
            data_var[:] = data.cpu().numpy()

output_path = "/root/autodl-tmp/processed_data.nc"
save_to_netcdf(output_path, gpu_data, new_lat, new_lon, time_tensor[:batch_size])

print(f"数据已成功保存到 {output_path}")
```
